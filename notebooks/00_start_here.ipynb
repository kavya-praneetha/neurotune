Three ways to call a model: **local (Ollama, free/offline)**, **Claude API**, and a **tool-using agent**.

## 1. Local model via Ollama — no API key, runs on your CPU

```python
from openai import OpenAI
ollama = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')
r = ollama.chat.completions.create(
    model='llama3.2:3b',
    messages=[{'role':'user','content':'Explain an AI agent in 2 sentences.'}])
print(r.choices[0].message.content)
```

## 2. Claude API
Put your key in `.env` (`ANTHROPIC_API_KEY=...`), then:

```python
from dotenv import load_dotenv; load_dotenv()
import anthropic
client = anthropic.Anthropic()
msg = client.messages.create(
    model='claude-sonnet-5', max_tokens=200,
    messages=[{'role':'user','content':'Give me a 1-line haiku about local LLMs.'}])
print(msg.content[0].text)
```

## 3. A tool-using agent (PydanticAI + Claude)
The model decides when to call your Python functions. See `agents/hello_agent.py` for the full version.

```python
from pydantic_ai import Agent
agent = Agent('anthropic:claude-sonnet-5',
              system_prompt='Use tools when relevant.')
@agent.tool_plain
def sqrt(x: float) -> float:
    'Return the square root of x.'
    return x ** 0.5
print(agent.run_sync('What is the square root of 20736?').output)
```
